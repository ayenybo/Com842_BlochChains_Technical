```python
avg_tx_size = 580        # bytes
block_time = 576         # seconds
block_sizes_mb = [1, 2, 4]

blocks_per_day = 24 * 60 * 60 / block_time    # 86,400 seconds in a day

print(f"{'Block size':<12}{'Tx per block':>14}{'TPS':>8}{'vs 1 MB':>10}{'Storage/year':>15}")
print("-" * 59)

base_tps = None
for mb in block_sizes_mb:
    size_bytes = mb * 1_000_000
    tx_per_block = size_bytes // avg_tx_size
    tps = tx_per_block / block_time
    if base_tps is None:
        base_tps = tps
    storage_gb_year = mb * blocks_per_day * 365 / 1000     # MB -> GB
    print(f"{str(mb) + ' MB':<12}{tx_per_block:>14,}{tps:>8.2f}{tps / base_tps:>9.2f}x{storage_gb_year:>12.2f} GB")

print()
print(f"Blocks per day = 86,400 / {block_time} = {blocks_per_day:.0f}")
```

```python
seconds_per_year = 365 * 24 * 60 * 60          # 31,536,000

btc_tps = {"1 MB": 1724 / 576, "2 MB": 3448 / 576, "4 MB": 6896 / 576}


visa_lecture_tps = 24_000                        # Week 5 lecture, slide 17
visa_capacity_tps = 83_000                       # Visa (2025): capacity, messages per second
visa_tx_per_year = 322_000_000_000               # Visa (2025): 322+ billion transactions a year
visa_avg_tps = visa_tx_per_year / seconds_per_year

print(f"Visa average TPS = {visa_tx_per_year:,} / {seconds_per_year:,} = {visa_avg_tps:,.0f}")
print()
print(f"{'Visa figure':<30}{'TPS':>10}{'x Bitcoin 1 MB':>16}{'x Bitcoin 4 MB':>16}")
print("-" * 72)
for label, v in [("Lecture figure (slide 17)", visa_lecture_tps),
                 ("Average (322B tx / year)", visa_avg_tps),
                 ("Peak capacity (Visa 2025)", visa_capacity_tps)]:
    print(f"{label:<30}{v:>10,.0f}{v / btc_tps['1 MB']:>15,.0f}x{v / btc_tps['4 MB']:>15,.0f}x")

print()
confirmations = 6
print(f"Bitcoin 6-confirmation wait = {confirmations} x 576 s = {confirmations * 576:,} s "
      f"= {confirmations * 576 / 60:.1f} minutes")
```

```python
import math

max_block_weight = 4_000_000     # BIP141 limit, in weight units (WU)
avg_tx_size = 580                # bytes (from the brief)
block_time = 576                 # seconds


witness_shares = [0.00, 0.25, 0.50, 0.60]

print(f"{'Witness share':<15}{'Weight (WU)':>12}{'vsize':>8}{'Tx/block':>10}{'TPS':>8}")
print("-" * 53)
for share in witness_shares:
    witness = avg_tx_size * share
    base = avg_tx_size - witness
    weight = base * 4 + witness
    vsize = math.ceil(weight / 4)
    tx_per_block = int(max_block_weight // weight)
    tps = tx_per_block / block_time
    print(f"{share:<15.0%}{weight:>12,.0f}{vsize:>8}{tx_per_block:>10,}{tps:>8.2f}")
```
